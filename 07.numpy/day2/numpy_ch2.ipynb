##### 다차원 배열 연산 : 요소별 연산

```python
A = np.array([[1,2,3],
             [4,5,6]])
B = np.array([[7,8,9],
              [10,11,12]])

print('A + B')
print(A + B)

print('A * B')
print(A * B)
```

##### 내적(dot product)

```python
# 1차원 벡터의 내적 : 두 배열을 곱하여 하나의 숫자로 만드는 연산
a = np.array([1,2,3])
b = np.array([4,5,6])

c = np.dot(a,b)
print(c)
```

```python
# 2차원 행렬의 곱(dot(), @연산, matmul() 연산 결과 동일)
A = np.array([[1,2],
              [3,4]])
B = np.array([[5,6],
              [7,8]])

print('np.dot(A,B)')
print(A@B)

print('np.matmul(A,B)')
print(np.matmul(A,B))
```

##### 3차원 배열 dot(a, b)과 matmul(a, b) 차이점 : dot()

```python
A = np.array([[[1,2],
               [3,4]],
               [[5,6],
               [7,8]]])
B = np.array([[[9,10],
               [11,12]],
               [[13,14],
                [15,16]]])
print('np.dot(A,B)')
print(np.dot(A,B))
```

##### 3차원 배열 dot(a, b)과 matmul(a, b) 차이점: matmul()

```python
A = np.array([[[1,2],
               [3,4]],
               [[5,6],
               [7,8]]])
B = np.array([[[9,10],
               [11,12]],
               [[13,14],
                [15,16]]])
print('np.matmul(A,B)')
print(np.matmul(A,B))
print('-'*10)

print('A @ B')
print(A @ B)
```

##### 외적(Outer Product)

```python
# 벡터의 외적
a = np.array([1,2,3])
b = np.array([4,5,6])    # 벡터(1차원 배열)

# 외적 계산                         (벡터가 아래처럼 2차원으로 변함)
outer_product = np.outer(a,b)    # 2차원 배열
print('외적 결과 :')
print(outer_product)
```

##### 2차원 배열의 전치 행렬 만들기

```python
A = np.array([[1,2,3],
              [4,5,6]])
print(A)    # A.shape : (2,3)
print('-' * 20)

# 방법 1 : .T 속성
print('전치행렬 A.T')
trans_a = A.T
print(trans_a)
print(f'A.T.shape : {trans_a.shape}')
print('-'*20)

# 방법 2 : transpose() 함수
print('np.transpose(A)')
print(np.transpose(A))
```

##### 역행렬(Inverse Matrix) 예제

```python
matrix = np.array([[1,2],
                   [3,4]])

# 역행렬 계산
inv_matrix = np.linalg.inv(matrix)
print('원본 행렬 :')
print(matrix)
print('\n역행렬 : ')
print(inv_matrix)

# 역행렬이 존재하는지 확인(행렬식이 0이 아님)
det_matrix = np.linalg.det(matrix)
print(f'\n행렬식 : {det_matrix}')
if det_matrix !=0:
    print('역행렬이 존재합니다!')
else:
    print('역행렬이 존재하지 않습니다.')
```

##### 배열의 정렬

```python
# 1차원 배열 정렬
arr = np.array([4,2,6,5,1,3,0])

# 정렬 (원본 유지)
sorted_arr = np.sort(arr)
print(f'정렬된 배열 : {sorted_arr}')    # [0 1 2 3 4 5 6]

# 원본 변경
arr.sort()
print(f'원본변경: {arr}')    # [0 1 2 3 4 5 6]

# 내림차순
desc_arr = np.sort(arr)[::-1]
print(f'내림차순 : {desc_arr}')    # [6 5 4 3 2 1 0]
print(f'원본배열 : {arr}')
```

```python
# 2차원 배열 정렬
arr2d = np.array([[3,1,6],
                  [0,7,4],
                  [5,3,2]])

# 행별 정렬 (axis=1)
print('sort(axis=1)')
row_sorted = np.sort(arr2d, axis=1)
print(row_sorted)

# 열별 정렬 (axis=0)
print('sort(axis=0)')
col_sorted = np.sort(arr2d, axis=0)
print(col_sorted)

# axis=None (1차원 배열로 변경 후 정렬)
print('sort(axis=None)')
none_sort = np.sort(arr2d, axis=None)
print(none_sort)
```

##### 정렬된 원소의 이전 인덱스 구하기

```python
arr = np.array([4,2,3,5,1])
print(f'원본 배열 : {arr}')

# 배열을 정렬할 때, 각 원소가 어디에 위치해야 되는지 알려줌
indicies = np.argsort(arr)
print(f'정렬시 원소 인덱스 : {indicies}')

# incidies로 배열 정렬 가능
print(f'배열 정렬 : {arr[indicies]}')

# 배열 정렬 : np.sort()
print(f'배열 정렬 : {np.sort(arr)}')
```

##### argsort() 활용: 기준 배열을 이용한 다른 배열 정렬

```python
fruit_array = np.array(['Apple', 'Banana', 'Cherry', 'Grape', 'Kiwi'])
price_array = np.array([800,1000,900,950,700])

desc_price = np.sort(price_array)[::-1]
print(f'price 내림차순 정렬 : {desc_price}')

# 정렬 인덱스 저장 : [::-1]를 붙일 경우 내림차순
# [::-1] :  내림차순으로 정렬할 때의 인덱스를 구함 => 가격에 매칭되는 과일 이름을 가져오기 위함
price_indicies = np.argsort(price_array)[::-1]    # price_indicies : [1 3 2 0 4]
print(fruit_array[price_indicies])    # 정렬된 가격 순서대로 fruit을 출력
```

##### 배열 병합 : vstack(), hstack()

```python
a = np.array([[1,2],
              [3,4]])
b = np.array([[5,6],
              [7,8]])

# 수직 병합 (행 추가)
vstack_result = np.vstack([a,b])
print(vstack_result)
print('-'*20)

# 수평 병합 (열 추가)
hstack_result = np.hstack([a,b])
print(hstack_result)
```

##### 배열 병합 : 1차원 배열 dstack()

```python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])

result = np.dstack((arr1, arr2))    # 튜플 전달
print(result)    # 출력 결과가 이해하기 힘듬. 강의자료 잘 참고할 것(헷갈리는 형태임)
print(result.shape)    # (1,3,2)
```

##### 배열 병합 : concatenate() 함수

```python
a = np.array([[1,2],
              [3,4]])
b = np.array([[5,6],
              [7,8]])

# axis=0 (수직)
concat_0 = np.concatenate([a,b], axis=0)
print(concat_0)
print()

# axis=1 (수평)
concat_1 = np.concatenate([a,b], axis=1)
print(concat_1)
print()

# axis=None (1차원 배열로 평탄화)
concat_none = np.concatenate([a,b], axis=None)
print(concat_none)
```

##### 배열 복사 방법 : 복사와 뷰 (Copy vs. View) 예제

```python
arr1 = np.array([1,2,3,4,5])

# view() : 얕은 복사(메모리 공유)
view_arr = arr1.view()
view_arr[0] = 100    # view_arr 수정

print(f'원본 배열(arr1): {arr1}')    # 원본 변경
print(f'view 배열(view_arr): {view_arr}')

# copy() : 깊은 복사(독립적인 메모리)
arr2 = np.array([1,2,3,4,5])
copy_arr = arr2.copy()
copy_arr[0] = 100    # copy_arr 수정

print(f'원본 배열(arr2): {arr2}')    # 원본 변경 안됨
print(f'copy 배열(copy_arr): {copy_arr}')    # copy_arr만 변경
```

##### 유니버설 함수 : 단항 유니버설 함수 #1

```python
arr = np.array([-2.5,-1.2,0,1.7,3.4])

print(f'np.ceil(): {np.ceil(arr)}')    # 가까운 정수로 올림
print(f'np.floor(): {np.floor(arr)}')    # 내림
print(f'np.rint(): {np.rint(arr)}')    # 가장 가까운 정수로 반올림(부동소수점 형태)

print(f'np.round(): {np.round(arr,0)}')    # 지정된 자리수로 반올림(정수 또는 부동소수점)

print(f'np.abs(): {np.abs(arr)}')    # 절대값
print(f'np.sqrt(): {np.sqrt(np.abs(arr))}')    # 제곱근

# 삼각함수
angles = np.array([0,np.pi/4, np.pi/2, np.pi])
print(np.sin(angles))
print(np.cos(angles))
```

##### 유니버설 함수 : 단항 유니버설 함수 #2

```python
x = np.arange(1,6)    # [1,2,3,4,5]

# reduce(x)
y = np.add.reduce(x)
print(y)

y = np.multiply.reduce(x)
print(y)

# accumulate(x) 계산 중 결과를 누적해서 저장
print(x)
y = np.add.accumulate(x)
print(y)
```

##### 유니버설 함수 : reduceat() 함수 #1

```python
arr = np.array([1,2,3,4,5,6])
indicies = [0,3]    # [0,3) : 1+2+3, [3:] : 4+5+6
# 0,3 : 구간의 시작 인덱스, ) : 불포함, [ : 포함

result = np.add.reduceat(arr,indicies)
print(result)    # [6 15]

result = np.multiply.reduceat(arr, [0,2,4])
print(result)
```

##### 유니버설 함수 : reduceat() 함수 #2

```python
arr = np.arange(1,13).reshape(4,3)
print('원본 배열:')
print(arr)

# 0~2행, 2~4행을 아래로 합침(indicies=[0,2])
result1 = np.add.reduceat(arr,[0,2], axis=0)
print(result1)

# 각 행 별로 첫 번째와 두 번째 열, 나머지 열을 합산
result2 = np.add.reduceat(arr, [0,2], axis=1)
print(result2)
```

##### ##### 유니버설 함수 : reduceat() 함수 #3

```python
arr = np.arange(10).reshape(2,5)
print(arr)
# incidies = [0,2,4] -> [0:2], [2:4], [4:]
result = np.multiply.reduceat(arr, [0,2,4], axis=1)
print(result)
print('-'*20)

arr = np.arange(1,17).reshape(4,4)
print(arr)
# 열 방향으로 구간합, 그 후 행 방향으로 또 reduceat 적용
result = np.add.reduceat(arr,[0,2], axis=1)
print(result)
final = np.add.reduceat(result, [0,2], axis=0)
print(final)
```

##### 유니버설 함수 : 이항 유니버설 함수 #1

##### PIL(Pillow) 및 Numpy를 활용한 간단한 이미지 처리 #1

```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

# 이미지 읽기 : Image.open('파일명')
img = Image.open('landscape.jpg')
img.show()    # PIL 사용

plt.imshow(img)
plt.show()    # matplotlib 사용
```

##### PIL(Pillow) 및 Numpy를 활용한 간단한 이미지 처리 #2

```python
img_array = np.array(img)    # PIL 이미지를 numpy 배열로 변환
print(img_array.shape)    # 컬러이미지 : (높이, 너비, 채널수)

# 1. RGB 채널 분리
r, g, b = img_array[:,:,0], img_array[:,:,1], img_array[:,:,2]    # 셋 다 각각 2차원 배열

# 2.1 가중치 공식을 적용하여 흑백 변환 : 시각적 보정(green 값을 높게 설정)
gray_array = (0.2989 * r + 0.5870 * g + 0.1140 * b).astype(np.uint8)

# 2.2 모든 색상을 동일한 비중으로 처리 (R+G+B)/3
#gray_array = np.mean(img_array, axis=2).astype(np.unit8)

# 2.3 PIL.convert('L') 사용하여 흑백 이미지 생성
gray_img = img.convert('L')
gray_img.show()

# 3. numpy 배열을 이용한 화면에 출력
# cmap='gray'를 사용하지 않으면  기본값 'viridis' 컬러맵으로 출력됨
plt.imshow(gray_array, cmap='gray')
plt.axis('off')
plt.show()
```
