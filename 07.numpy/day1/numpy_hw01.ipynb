#### 4명의 학생과 4개의 과목의 성적 데이터 분석

```python
import numpy as np

subjects = np.array(['수학', '영어', '과학', '사회'])
student_data = np.array([
    ['김철수', 90, 85, 78, 92],
    ['이영희', 76, 80, 79, 85],
    ['박민수', 88, 92, 91, 86],
    ['최지연', 70, 75, 80, 85],
])
```

##### * 학생 이름과 점수를 분리하여 아래와 같이 화면에 출력

```python
names = student_data[:,0]
scores = np.array(student_data[:,1:], dtype='int32')
print(f'학생 이름:{names}')
print('점수 데이터:')
print(f'{scores}')
```

##### * 과목별 평균을 계산하고 가장 평균이 높은 과목을 화면에 출력

```python
avg_scores = np.mean(scores, axis=0)
print('과목별 평균 점수')
for i, avg in enumerate(avg_scores):
    print(f'{subjects[i]} : {avg:.1f}점')

high_avg = np.max(avg_scores)
high_sub = subjects[np.argmax(avg_scores)]
print()
print(f'평균이 가장 높은 과목 : {high_sub} ({high_avg}점)')
```

##### * 학생별 총점과 평균을 계산하고 총점이 가장 높은 학생의 이름 및 총점 출력

```python
print('학생별 성적')
for i, avg in enumerate(scores):
    print(f'{names[i]} : 총점 {np.sum(scores[i])}점, 평균 {np.mean(scores[i]):.1f}점')

total = np.sum(scores, axis=1)
high_total = np.max(total)
high_student = names[np.argmax(total)]
mean = np.mean(scores, axis=1)
high_mean = np.max(mean)
print()
print(f'1등 학생 : {high_student} (총점 : {high_total}점, 평균 : {high_mean:.1f}점)')
```

##### * 각 과목별 80점 이상인 점수 찾기(Boolean indexing)

```python
above80 = scores >= 80
print('80점 이상 과목 :')
print(f'{above80}')
```

##### * 학생별 80점 이상인 과목의 수 출력

```python
count_over80 = np.sum(above80, axis=1)
print('학생별 80점 이상 과목 수 :')
for i in range(len(names)):
    print(f'{names[i]} : {count_over80[i]}과목')
```

##### * 모든 과목의 점수가 80점 이상인 학생 출력 및 수학 과목이 90점인 학생 이름 출력

```python
all_above_80 = np.all(scores >= 80, axis=1)
math_over_90 = scores[:, 0] >= 90
student_high = np.where(all_above_80)[0]
student_math = np.where(math_over_90)[0]
print('모든 과목이 80점 이상인 학생 :', names[student_high][0])
print('수학 90점 이상인 학생 :', names[student_math][0])
```

##### * 전체 점수 통계(전체 평균, 최고점수, 최저점수, 표준편차 계산)

```python
print('=== 전체 점수 통계 ===')
print(f'전체 평균 : {np.mean(scores):.1f}점')
print(f'최고 점수 : {np.max(scores)}점')
print(f'최저 점수 : {np.min(scores)}점')
print(f'표준편차 : {np.std(scores):.1f}점')
```

##### * 학생 별 최고/최저 점수 과목 출력

```python
high_score_sub = np.max(scores, axis=0)
min_score_sub = np.min(scores, axis=0)
high_score_stu = np.max(scores, axis=1)
min_score_stu = np.min(scores, axis=1)
print('=== 학생별 최고/최저 과목 ===')
for i in range(len(names)):
    print(names[i])
    print(f'   최고 점수 : {high_score_stu[i]}점')
    print(f'   최저 점수 : {min_score_stu[i]}점')
```

##### * 과목별 점수 분포 출력(최고 점수, 최저 점수, 점수 차이(최고점수-최저점수)

```python
print('=== 과목별 점수 분포 ===')
for i in range(len(subjects)):
    print(f'{subjects[i]}: 최고 : {high_score_sub[i]}점, 최저 : {min_score_sub[i]}점, 점수차이 : {high_score_sub[i] - min_score_sub[i]}점')
```
