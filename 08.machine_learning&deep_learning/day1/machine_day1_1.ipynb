### 머신러닝의 개요 및 KNN
##### KDT 12기 이재헌

```python
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
```

```python
# 도미 데이터

bream_length = [25.4, 26.3, 26.5, 29.0, 29.0, 29.7, 29.7, 30.0, 30.0, 30.7, 31.0, 31.0, 
                31.5, 32.0, 32.0, 32.0, 33.0, 33.0, 33.5, 33.5, 34.0, 34.0, 34.5, 35.0, 
                35.0, 35.0, 35.0, 36.0, 36.0, 37.0, 38.5, 38.5, 39.5, 41.0, 41.0]
bream_weight = [242.0, 290.0, 340.0, 363.0, 430.0, 450.0, 500.0, 390.0, 450.0, 500.0, 475.0, 500.0, 
                500.0, 340.0, 600.0, 600.0, 700.0, 700.0, 610.0, 650.0, 575.0, 685.0, 620.0, 680.0, 
                700.0, 725.0, 720.0, 714.0, 850.0, 1000.0, 920.0, 955.0, 925.0, 975.0, 950.0]

plt.scatter(bream_length, bream_weight)
plt.xlabel('length')    # x축은 길이
plt.ylabel('weight')    # y축은 무게
plt.show()
```

```python
# 빙어 데이터

smelt_length = [9.8, 10.5, 10.6, 11.0, 11.2, 11.3, 11.8, 11.8, 12.0, 12.2, 12.4, 13.0, 14.3, 15.0]
smelt_weight = [6.7, 7.5, 7.0, 9.7, 9.8, 8.7, 10.0, 9.9, 9.8, 12.2, 13.4, 12.2, 19.7, 19.9]

plt.scatter(bream_length, bream_weight)    # 파란색
plt.scatter(smelt_length, smelt_weight)    # 주황색
plt.xlabel('length')
plt.ylabel('weight')
plt.show()
```

```python
# 첫 번째 머신러닝 프로그램

length = bream_length + smelt_length
weight = bream_weight + smelt_weight

# 도미와 빙어 데이터를 합쳐서 2차원 리스트 만들기
fish_data = [[l,w] for l,w in zip(length, weight)]
print(fish_data)

fish_target = [1] * 35 + [0] * 14    # 1은 도미(35마리), 0은 빙어(14마리)
print(fish_target)
```

```python
# import한 KNeighborsClassifier 클래스의 객체 만들기
kn = KNeighborsClassifier()

# 훈련(training)
kn.fit(fish_data, fish_target)
```

```python
# 성능평가
kn.score(fish_data, fish_target)    # 여기서 1.0 나옴
```

```python
plt.scatter(bream_length, bream_weight)    # 파란색
plt.scatter(smelt_length, smelt_weight)    # 주황색
plt.scatter(30,600, marker='*', color='tomato', s=50)    # 새로운 샘플 데이터
plt.xlabel('length')
plt.ylabel('weight')
plt.show()
```

```python
kn.predict([[30,600]])    # 도미로 예측됨
```

```python
kn.predict([[14,150]])    # 빙어로 예측됨
```

```python
kn49 = KNeighborsClassifier(n_neighbors=49)    # 참고 데이터를 49개로 한 kn49 모델
# 하이퍼 파라미터

# 훈련
kn49.fit(fish_data, fish_target)
```

```python
# 평가
kn49.score(fish_data, fish_target)    # k값이 너무 커서 과소적합
```

```python
print(35/49)
```

```python
kn = KNeighborsClassifier()
kn.fit(fish_data, fish_target)
for n in range(5,50):
    # k-최근접 이웃 개수 설정
    kn.n_neighbors = n
    score = kn.score(fish_data, fish_target)

    if score < 1:
        print(n, score)
        break
```
